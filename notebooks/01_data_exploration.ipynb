## Narrowing the window

The full log covers Sep 2018 to Sep 2020, about 31.8M rows. The analysis uses the final three months (Jun 22 to Sep 22, 2020), which leaves 4,056,792 rows. That keeps the load and the analysis manageable. It also creates the left-censoring effect discussed in notebook 03.

```python
cutoff_date = transactions['t_dat'].max() - pd.DateOffset(months=3)
transactions_filtered = transactions[transactions['t_dat'] >= cutoff_date].copy()
print("Filtered shape:", transactions_filtered.shape)
```

## Sampling customers, not rows

Half of the customers active in the window are sampled (seed 42). Every transaction for a sampled customer is kept, so each customer's purchase history stays complete. Sampling individual rows would break recency, frequency and cohort membership.

Result: 267,708 customers and 2,029,325 transactions.

```python
import numpy as np

np.random.seed(42)  # keeps your sample reproducible if you rerun this later

sample_fraction = 0.5  # start here, adjust based on resulting file size
sampled_customer_ids = np.random.choice(
    transactions_filtered['customer_id'].unique(),
    size=int(len(transactions_filtered['customer_id'].unique()) * sample_fraction),
    replace=False
)

transactions_sampled = transactions_filtered[
    transactions_filtered['customer_id'].isin(sampled_customer_ids)
].copy()

print("Sampled shape:", transactions_sampled.shape)
```

```python
print("Sample fraction used:", sample_fraction)
print("Number of sampled customers:", len(sampled_customer_ids))
print("Transactions sampled shape:", transactions_sampled.shape)
```

```python
transactions_sampled.to_csv('../data/processed/transactions_filtered.csv', index=False)
```

## Loading into Supabase

The connection string lives in a local `.env` file that is not committed. Supabase's direct connection is IPv6-only, so the Session pooler connection string is used instead. Loading uses `copy_expert` streaming, because the browser CSV importer ran out of memory partway through.

```python
from dotenv import load_dotenv
import psycopg2
import os

conn = psycopg2.connect(os.getenv("SUPABASE_CONNECTION_STRING"))
cur = conn.cursor()
print("Connected successfully")
```

After any failed statement, PostgreSQL rejects everything that follows in that transaction (`InFailedSqlTransaction`) until you roll back.

```python
conn.rollback()
print("Transaction rolled back")
```

## Foreign keys: drop, load, rebuild

With the foreign keys in place, the bulk load hit a statement timeout from validating each row as it arrived. Dropping the two constraints, loading, then adding them back validates every row against `customers` and `articles` in one pass instead.

```python
cur.execute("ALTER TABLE transactions DROP CONSTRAINT transactions_customer_id_fkey;")
cur.execute("ALTER TABLE transactions DROP CONSTRAINT transactions_article_id_fkey;")
conn.commit()
print("Constraints dropped")
```

```python
cur.execute("TRUNCATE TABLE transactions;")
conn.commit()

with open('../data/processed/transactions_filtered.csv', 'r', encoding='utf-8') as f:
    next(f)
    cur.copy_expert(
        "COPY transactions (t_dat, customer_id, article_id, price, sales_channel_id) FROM STDIN WITH CSV",
        f
    )
conn.commit()
print("Import complete")
```

```python
cur.execute("ALTER TABLE transactions ADD CONSTRAINT transactions_customer_id_fkey FOREIGN KEY (customer_id) REFERENCES customers(customer_id);")
cur.execute("ALTER TABLE transactions ADD CONSTRAINT transactions_article_id_fkey FOREIGN KEY (article_id) REFERENCES articles(article_id);")
conn.commit()
print("Constraints restored")
```
